```python
from pathlib import Path
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import rasterio
from rasterio.warp import reproject, Resampling
from pyproj import Transformer

# Settings
LANDSAT_FOLDER = Path("../scratch/optical/landsat_oursites_L1_GEE/TOA/")
ELEVATION_FOLDER = Path("../scratch/optical/landsat_masking/")
WC_FOLDER = Path("../scratch/optical/WC/")

CELL_SIZE_M = 200
DISTANCE_BIN_M = 200
MAX_DISTANCE_M = None
MAX_ELEVATION_DIFFERENCE_M = 80
MIN_VALID_FRACTION = 0.20
PERCENTILES = (16, 84)

# Defaults to the image centre. Optional overrides: camera: (longitude, latitude).
CAMERA_LONLAT = {}

CAMERAS = [
    "bartlett", "delta_junction", "glees", "hyytiala", "kenttarova",
    "lacclair", "marcell", "old_jack_pine", "oregon", "queens",
    "sodankyla", "tammela", "torgnon", "underc", "underhill",
    "varrio", "willowcreek", "wslcreek",
]
WC_NAME_MAP = {
    "sodankyla": "sodankyla_full",
    "marcell": "marcell_MN",
    "oregon": "oregon_yp",
}

assert CELL_SIZE_M > 0 and DISTANCE_BIN_M > 0
assert 0 <= MIN_VALID_FRACTION <= 1

landsat_files = sorted(LANDSAT_FOLDER.glob("*.tif"))
if not landsat_files:
    raise FileNotFoundError(f"No Landsat files found in {LANDSAT_FOLDER}")

elevation_files = list(ELEVATION_FOLDER.iterdir())
cell_results = []

for filepath in landsat_files:
    # Observed FSC is encoded in the filename, as in your original script.
    match = re.search(rf"({'|'.join(CAMERAS)})(\d+)_", filepath.name)
    if match is None:
        continue

    camera = match.group(1)
    observed_fsc = int(match.group(2))

    if not 0 < observed_fsc < 100:
        continue

    elevation_matches = [
        p for p in elevation_files
        if camera in p.name
        and "elevation" in p.name.lower()
        and p.suffix.lower() == ".tif"
    ]
    if len(elevation_matches) != 1:
        raise ValueError(
            f"{camera}: expected one elevation file, "
            f"found {len(elevation_matches)}."
        )

    wc_path = WC_FOLDER / f"{WC_NAME_MAP.get(camera, camera)}.tif"
    print(f"Processing {filepath.name}")

    with rasterio.open(filepath) as src:
        if (
            src.crs is None
            or not src.crs.is_projected
            or not np.isclose(src.crs.linear_units_factor[1], 1)
        ):
            raise ValueError(f"{filepath.name}: CRS must use projected metres.")

        transform = src.transform
        bounds = src.bounds
        height, width = src.height, src.width

        if (
            transform.b != 0 or transform.d != 0
            or transform.a <= 0 or transform.e >= 0
        ):
            raise ValueError(f"{filepath.name}: expected a north-up raster.")

        if CELL_SIZE_M < max(transform.a, -transform.e):
            raise ValueError("CELL_SIZE_M must be at least the pixel size.")

        if camera in CAMERA_LONLAT:
            cx, cy = Transformer.from_crs(
                "EPSG:4326", src.crs, always_xy=True
            ).transform(*CAMERA_LONLAT[camera])
        else:
            cx = (bounds.left + bounds.right) / 2
            cy = (bounds.bottom + bounds.top) / 2

        row, col = src.index(cx, cy)
        if not (0 <= row < height and 0 <= col < width):
            raise ValueError(f"{camera}: camera is outside {filepath.name}.")

        # Green, NIR and SWIR1 bands; reflectance is already scaled.
        if "Landsat7" in filepath.name:
            bands = [2, 4, 5]
        elif "Landsat8" in filepath.name:
            bands = [3, 5, 6]
        else:
            raise ValueError(f"Unknown sensor: {filepath.name}")

        green, nir, swir = (
            src.read(bands, masked=True)
            .astype("float32")
            .filled(np.nan)
        )

        # Reproject forest and elevation layers onto the Landsat grid.
        masks = {}
        for name, path, method in [
            ("forest", wc_path, Resampling.nearest),
            ("elevation", elevation_matches[0], Resampling.bilinear),
        ]:
            aligned = np.full((height, width), np.nan, dtype="float32")

            with rasterio.open(path) as mask_src:
                source = (
                    mask_src.read(1, masked=True)
                    .astype("float32")
                    .filled(np.nan)
                )
                reproject(
                    source=source,
                    destination=aligned,
                    src_transform=mask_src.transform,
                    src_crs=mask_src.crs,
                    src_nodata=np.nan,
                    dst_transform=transform,
                    dst_crs=src.crs,
                    dst_nodata=np.nan,
                    resampling=method,
                )

            masks[name] = aligned
            del source

    centre_elevation = masks["elevation"][row, col]
    if not np.isfinite(centre_elevation):
        print("  Skipping: missing camera elevation.")
        continue

    # Your original Dozier classification.
    denominator = green + swir
    ndsi = np.divide(
        green - swir,
        denominator,
        out=np.full_like(green, np.nan),
        where=denominator != 0,
    )
    snow = (
        ((ndsi > 0.1) & (ndsi < 0.4))
        | ((ndsi >= 0.4) & (nir > 0.11))
    )

    valid = (
        np.isfinite(ndsi)
        & np.isfinite(nir)
        & (masks["forest"] == 10)
        & (
            np.abs(masks["elevation"] - centre_elevation)
            <= MAX_ELEVATION_DIFFERENCE_M
        )
    )

    # Assign pixels to non-overlapping cells, one centred on the camera.
    x = transform.c + (np.arange(width) + 0.5) * transform.a
    y = transform.f + (np.arange(height) + 0.5) * transform.e

    gx = np.floor((x - cx) / CELL_SIZE_M + 0.5).astype(int)
    gy = np.floor((y - cy) / CELL_SIZE_M + 0.5).astype(int)

    nx = gx.max() - gx.min() + 1
    ny = gy.max() - gy.min() + 1
    cell_id = (
        (gy[:, None] - gy.min()) * nx
        + (gx[None, :] - gx.min())
    )

    total_count = np.bincount(cell_id.ravel(), minlength=nx * ny)
    valid_count = np.bincount(cell_id[valid], minlength=nx * ny)
    snow_count = np.bincount(cell_id[valid & snow], minlength=nx * ny)

    grid_x, grid_y = np.meshgrid(
        np.arange(gx.min(), gx.max() + 1),
        np.arange(gy.min(), gy.max() + 1),
    )

    cells = pd.DataFrame({
        "grid_x": grid_x.ravel(),
        "grid_y": grid_y.ravel(),
        "n_valid_pixels": valid_count,
        "valid_fraction": valid_count / np.maximum(total_count, 1),
        "fsc_pct": np.divide(
            100.0 * snow_count,
            valid_count,
            out=np.full(nx * ny, np.nan),
            where=valid_count > 0,
        ),
    })

    cells["x"] = cx + cells["grid_x"] * CELL_SIZE_M
    cells["y"] = cy + cells["grid_y"] * CELL_SIZE_M
    cells["distance_m"] = CELL_SIZE_M * np.hypot(
        cells["grid_x"], cells["grid_y"]
    )

    # Retain complete cells with enough valid forest pixels.
    half = CELL_SIZE_M / 2
    keep = (
        (cells["x"] - half >= bounds.left)
        & (cells["x"] + half <= bounds.right)
        & (cells["y"] - half >= bounds.bottom)
        & (cells["y"] + half <= bounds.top)
        & (cells["valid_fraction"] >= MIN_VALID_FRACTION)
        & cells["fsc_pct"].notna()
    )
    if MAX_DISTANCE_M is not None:
        keep &= cells["distance_m"] <= MAX_DISTANCE_M

    cells = cells.loc[keep].copy()

    # Release full-scene arrays before processing the next scene.
    del green, nir, swir, denominator, ndsi, snow, valid
    del masks, aligned, cell_id

    centre = cells.loc[
        (cells["grid_x"] == 0) & (cells["grid_y"] == 0), "fsc_pct"
    ]
    if centre.empty:
        print("  Skipping: no usable central FSC cell.")
        continue

    cells["centre_fsc_pct"] = centre.iloc[0]
    cells["difference_pp"] = cells["fsc_pct"] - centre.iloc[0]
    cells["observed_fsc_pct"] = observed_fsc
    cells["camera"] = camera
    cells["scene"] = filepath.name
    cell_results.append(cells)

if not cell_results:
    raise ValueError("No partial-snow scenes with usable central FSC cells.")

cells_df = pd.concat(cell_results, ignore_index=True)
del cell_results

# Plot 1: one point per scene, coloured by location.
scenes_df = cells_df[
    ["scene", "camera", "observed_fsc_pct", "centre_fsc_pct"]
].drop_duplicates("scene")

fig, ax = plt.subplots(figsize=(6, 6))
locations = sorted(scenes_df["camera"].unique())
colours = plt.get_cmap("tab20", len(locations))

for i, location in enumerate(locations):
    subset = scenes_df.loc[scenes_df["camera"] == location]
    ax.scatter(
        subset["observed_fsc_pct"],
        subset["centre_fsc_pct"],
        color=colours(i),
        label=location,
        s=40,
        alpha=0.8,
        edgecolors="none",
    )

ax.plot([0, 100], [0, 100], "--", color="grey", lw=1)
ax.set(
    xlabel="Observed FSC (%)",
    ylabel="Landsat FSC in camera-centred cell (%)",
    xlim=(-2, 102),
    ylim=(-2, 102),
)
ax.set_aspect("equal")
ax.grid(alpha=0.2)
ax.legend(
    title="Location", bbox_to_anchor=(1.02, 1), loc="upper left"
)
fig.tight_layout()
plt.show()

# Plot 2: pool differences across all scenes, giving each cell equal weight.
# Exclude central cells, whose differences are zero by definition.
outer_cells = cells_df.loc[cells_df["distance_m"] > 0].copy()
distance_summary = pd.DataFrame()

if outer_cells.empty:
    print("No usable cells outside the central cells.")
else:
    outer_cells["distance_bin"] = (
        outer_cells["distance_m"] // DISTANCE_BIN_M
    ).astype(int)

    distance_summary = (
        outer_cells.groupby("distance_bin")
        .agg(
            mean_difference_pp=("difference_pp", "mean"),
            lower_pp=(
                "difference_pp",
                lambda v: np.percentile(v, PERCENTILES[0]),
            ),
            upper_pp=(
                "difference_pp",
                lambda v: np.percentile(v, PERCENTILES[1]),
            ),
            n_cells=("difference_pp", "size"),
            n_scenes=("scene", "nunique"),
        )
    )

    # Preserve empty bins as gaps in the plot.
    distance_summary = distance_summary.reindex(
        range(int(outer_cells["distance_bin"].max()) + 1)
    )
    distance_summary["distance_m"] = (
        distance_summary.index + 0.5
    ) * DISTANCE_BIN_M

    fig, ax = plt.subplots(figsize=(8, 4.5))

    ax.fill_between(
        distance_summary["distance_m"].to_numpy(),
        distance_summary["lower_pp"].to_numpy(),
        distance_summary["upper_pp"].to_numpy(),
        alpha=0.25,
        label=f"{PERCENTILES[0]}–{PERCENTILES[1]}th percentiles",
    )
    ax.plot(
        distance_summary["distance_m"],
        distance_summary["mean_difference_pp"],
        "o-",
        ms=4,
        label="Mean difference",
    )
    ax.axhline(0, color="black", lw=0.8)
    ax.set(
        xlabel="Distance from camera (m)",
        ylabel="Cell FSC − central FSC (percentage points)",
    )
    ax.grid(alpha=0.2)
    ax.legend()
    fig.tight_layout()
    plt.show()
```

```python
from pathlib import Path
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import rasterio
from rasterio.warp import reproject, Resampling
from pyproj import Transformer

# Settings
LANDSAT_FOLDER = Path("../scratch/optical/landsat_oursites_L1_GEE/TOA/")
ELEVATION_FOLDER = Path("../scratch/optical/landsat_masking/")
WC_FOLDER = Path("../scratch/optical/WC/")

CELL_SIZE_M = 200
DISTANCE_BIN_M = 200
MAX_DISTANCE_M = None
MAX_ELEVATION_DIFFERENCE_M = 80
MIN_VALID_FRACTION = 0.20
PERCENTILES = (16, 84)
FSC_BIN_EDGES = [0, 20, 40, 60, 80, 100]  # Observed FSC (%).
SHOW_FSC_SPREAD = False  # Set True to shade each group's percentile range.

# Defaults to the image centre. Optional overrides: camera: (longitude, latitude).
CAMERA_LONLAT = {}

CAMERAS = [
    "bartlett", "delta_junction", "glees", "hyytiala", "kenttarova",
    "lacclair", "marcell", "old_jack_pine", "oregon", "queens",
    "sodankyla", "tammela", "torgnon", "underc", "underhill",
    "varrio", "willowcreek", "wslcreek",
]
WC_NAME_MAP = {
    "sodankyla": "sodankyla_full",
    "marcell": "marcell_MN",
    "oregon": "oregon_yp",
}

assert CELL_SIZE_M > 0 and DISTANCE_BIN_M > 0
assert 0 <= MIN_VALID_FRACTION <= 1

landsat_files = sorted(LANDSAT_FOLDER.glob("*.tif"))
if not landsat_files:
    raise FileNotFoundError(f"No Landsat files found in {LANDSAT_FOLDER}")

elevation_files = list(ELEVATION_FOLDER.iterdir())
cell_results = []

for filepath in landsat_files:
    # Observed FSC is encoded in the filename, as in your original script.
    match = re.search(rf"({'|'.join(CAMERAS)})(\d+)_", filepath.name)
    if match is None:
        continue

    camera = match.group(1)
    observed_fsc = int(match.group(2))

    if not 0 < observed_fsc < 100:
        continue

    elevation_matches = [
        p for p in elevation_files
        if camera in p.name
        and "elevation" in p.name.lower()
        and p.suffix.lower() == ".tif"
    ]
    if len(elevation_matches) != 1:
        raise ValueError(
            f"{camera}: expected one elevation file, "
            f"found {len(elevation_matches)}."
        )

    wc_path = WC_FOLDER / f"{WC_NAME_MAP.get(camera, camera)}.tif"
    print(f"Processing {filepath.name}")

    with rasterio.open(filepath) as src:
        if (
            src.crs is None
            or not src.crs.is_projected
            or not np.isclose(src.crs.linear_units_factor[1], 1)
        ):
            raise ValueError(f"{filepath.name}: CRS must use projected metres.")

        transform = src.transform
        bounds = src.bounds
        height, width = src.height, src.width

        if (
            transform.b != 0 or transform.d != 0
            or transform.a <= 0 or transform.e >= 0
        ):
            raise ValueError(f"{filepath.name}: expected a north-up raster.")

        if CELL_SIZE_M < max(transform.a, -transform.e):
            raise ValueError("CELL_SIZE_M must be at least the pixel size.")

        if camera in CAMERA_LONLAT:
            cx, cy = Transformer.from_crs(
                "EPSG:4326", src.crs, always_xy=True
            ).transform(*CAMERA_LONLAT[camera])
        else:
            cx = (bounds.left + bounds.right) / 2
            cy = (bounds.bottom + bounds.top) / 2

        row, col = src.index(cx, cy)
        if not (0 <= row < height and 0 <= col < width):
            raise ValueError(f"{camera}: camera is outside {filepath.name}.")

        # Green, NIR and SWIR1 bands; reflectance is already scaled.
        if "Landsat7" in filepath.name:
            bands = [2, 4, 5]
        elif "Landsat8" in filepath.name:
            bands = [3, 5, 6]
        else:
            raise ValueError(f"Unknown sensor: {filepath.name}")

        green, nir, swir = (
            src.read(bands, masked=True)
            .astype("float32")
            .filled(np.nan)
        )

        # Reproject forest and elevation layers onto the Landsat grid.
        masks = {}
        for name, path, method in [
            ("forest", wc_path, Resampling.nearest),
            ("elevation", elevation_matches[0], Resampling.bilinear),
        ]:
            aligned = np.full((height, width), np.nan, dtype="float32")

            with rasterio.open(path) as mask_src:
                source = (
                    mask_src.read(1, masked=True)
                    .astype("float32")
                    .filled(np.nan)
                )
                reproject(
                    source=source,
                    destination=aligned,
                    src_transform=mask_src.transform,
                    src_crs=mask_src.crs,
                    src_nodata=np.nan,
                    dst_transform=transform,
                    dst_crs=src.crs,
                    dst_nodata=np.nan,
                    resampling=method,
                )

            masks[name] = aligned
            del source

    centre_elevation = masks["elevation"][row, col]
    if not np.isfinite(centre_elevation):
        print("  Skipping: missing camera elevation.")
        continue

    # Your original Dozier classification.
    denominator = green + swir
    ndsi = np.divide(
        green - swir,
        denominator,
        out=np.full_like(green, np.nan),
        where=denominator != 0,
    )
    snow = (
        ((ndsi > 0.1) & (ndsi < 0.4))
        | ((ndsi >= 0.4) & (nir > 0.11))
    )

    valid = (
        np.isfinite(ndsi)
        & np.isfinite(nir)
        & (masks["forest"] == 10)
        & (
            np.abs(masks["elevation"] - centre_elevation)
            <= MAX_ELEVATION_DIFFERENCE_M
        )
    )

    # Assign pixels to non-overlapping cells, one centred on the camera.
    x = transform.c + (np.arange(width) + 0.5) * transform.a
    y = transform.f + (np.arange(height) + 0.5) * transform.e

    gx = np.floor((x - cx) / CELL_SIZE_M + 0.5).astype(int)
    gy = np.floor((y - cy) / CELL_SIZE_M + 0.5).astype(int)

    nx = gx.max() - gx.min() + 1
    ny = gy.max() - gy.min() + 1
    cell_id = (
        (gy[:, None] - gy.min()) * nx
        + (gx[None, :] - gx.min())
    )

    total_count = np.bincount(cell_id.ravel(), minlength=nx * ny)
    valid_count = np.bincount(cell_id[valid], minlength=nx * ny)
    snow_count = np.bincount(cell_id[valid & snow], minlength=nx * ny)

    grid_x, grid_y = np.meshgrid(
        np.arange(gx.min(), gx.max() + 1),
        np.arange(gy.min(), gy.max() + 1),
    )

    cells = pd.DataFrame({
        "grid_x": grid_x.ravel(),
        "grid_y": grid_y.ravel(),
        "n_valid_pixels": valid_count,
        "valid_fraction": valid_count / np.maximum(total_count, 1),
        "fsc_pct": np.divide(
            100.0 * snow_count,
            valid_count,
            out=np.full(nx * ny, np.nan),
            where=valid_count > 0,
        ),
    })

    cells["x"] = cx + cells["grid_x"] * CELL_SIZE_M
    cells["y"] = cy + cells["grid_y"] * CELL_SIZE_M
    cells["distance_m"] = CELL_SIZE_M * np.hypot(
        cells["grid_x"], cells["grid_y"]
    )

    # Retain complete cells with enough valid forest pixels.
    half = CELL_SIZE_M / 2
    keep = (
        (cells["x"] - half >= bounds.left)
        & (cells["x"] + half <= bounds.right)
        & (cells["y"] - half >= bounds.bottom)
        & (cells["y"] + half <= bounds.top)
        & (cells["valid_fraction"] >= MIN_VALID_FRACTION)
        & cells["fsc_pct"].notna()
    )
    if MAX_DISTANCE_M is not None:
        keep &= cells["distance_m"] <= MAX_DISTANCE_M

    cells = cells.loc[keep].copy()

    # Release full-scene arrays before processing the next scene.
    del green, nir, swir, denominator, ndsi, snow, valid
    del masks, aligned, cell_id

    centre = cells.loc[
        (cells["grid_x"] == 0) & (cells["grid_y"] == 0), "fsc_pct"
    ]
    if centre.empty:
        print("  Skipping: no usable central FSC cell.")
        continue

    cells["centre_fsc_pct"] = centre.iloc[0]
    cells["difference_pp"] = cells["fsc_pct"] - centre.iloc[0]
    cells["observed_fsc_pct"] = observed_fsc
    cells["camera"] = camera
    cells["scene"] = filepath.name
    cell_results.append(cells)

if not cell_results:
    raise ValueError("No partial-snow scenes with usable central FSC cells.")

cells_df = pd.concat(cell_results, ignore_index=True)
del cell_results

# Plot 1: one point per scene, coloured by location.
scenes_df = cells_df[
    ["scene", "camera", "observed_fsc_pct", "centre_fsc_pct"]
].drop_duplicates("scene")

fig, ax = plt.subplots(figsize=(6, 6))
locations = sorted(scenes_df["camera"].unique())
colours = plt.get_cmap("tab20", len(locations))

for i, location in enumerate(locations):
    subset = scenes_df.loc[scenes_df["camera"] == location]
    ax.scatter(
        subset["observed_fsc_pct"],
        subset["centre_fsc_pct"],
        color=colours(i),
        label=location,
        s=40,
        alpha=0.8,
        edgecolors="none",
    )

ax.plot([0, 100], [0, 100], "--", color="grey", lw=1)
ax.set(
    xlabel="Observed FSC (%)",
    ylabel="Landsat FSC in camera-centred cell (%)",
    xlim=(-2, 102),
    ylim=(-2, 102),
)
ax.set_aspect("equal")
ax.grid(alpha=0.2)
ax.legend(
    title="Location", bbox_to_anchor=(1.02, 1), loc="upper left"
)
fig.tight_layout()
plt.show()

# Plot 2: pool differences across all scenes, giving each cell equal weight.
# Exclude central cells, whose differences are zero by definition.
outer_cells = cells_df.loc[cells_df["distance_m"] > 0].copy()
distance_summary = pd.DataFrame()
fsc_distance_summary = pd.DataFrame()

if outer_cells.empty:
    print("No usable cells outside the central cells.")
else:
    outer_cells["distance_bin"] = (
        outer_cells["distance_m"] // DISTANCE_BIN_M
    ).astype(int)

    distance_summary = (
        outer_cells.groupby("distance_bin")
        .agg(
            mean_difference_pp=("difference_pp", "mean"),
            lower_pp=(
                "difference_pp",
                lambda v: np.percentile(v, PERCENTILES[0]),
            ),
            upper_pp=(
                "difference_pp",
                lambda v: np.percentile(v, PERCENTILES[1]),
            ),
            n_cells=("difference_pp", "size"),
            n_scenes=("scene", "nunique"),
        )
    )

    # Preserve empty bins as gaps in the plot.
    distance_summary = distance_summary.reindex(
        range(int(outer_cells["distance_bin"].max()) + 1)
    )
    distance_summary["distance_m"] = (
        distance_summary.index + 0.5
    ) * DISTANCE_BIN_M

    # Group scenes by observed FSC, pooling sites within each FSC range.
    # Bins include their upper edge: e.g. 20% belongs to the 0–20% group.
    fsc_edges = np.asarray(FSC_BIN_EDGES, dtype=float)
    if (
        len(fsc_edges) < 2 or not np.all(np.isfinite(fsc_edges))
        or not np.all(np.diff(fsc_edges) > 0)
        or fsc_edges[0] != 0 or fsc_edges[-1] != 100
    ):
        raise ValueError("FSC_BIN_EDGES must increase strictly from 0 to 100.")
    outer_cells["fsc_bin"] = pd.cut(
        outer_cells["observed_fsc_pct"], bins=fsc_edges,
        labels=False, include_lowest=True, right=True,
    )
    fsc_distance_summary = (
        outer_cells.groupby(["fsc_bin", "distance_bin"])
        .agg(
            mean_difference_pp=("difference_pp", "mean"),
            lower_pp=("difference_pp", lambda v: np.percentile(v, PERCENTILES[0])),
            upper_pp=("difference_pp", lambda v: np.percentile(v, PERCENTILES[1])),
            n_cells=("difference_pp", "size"),
            n_scenes=("scene", "nunique"),
            n_cameras=("camera", "nunique"),
        )
        .reset_index()
    )
    fsc_distance_summary["distance_m"] = (
        fsc_distance_summary["distance_bin"] + 0.5
    ) * DISTANCE_BIN_M
    fsc_distance_summary["fsc_lower_pct"] = fsc_edges[
        fsc_distance_summary["fsc_bin"].to_numpy(dtype=int)
    ]
    fsc_distance_summary["fsc_upper_pct"] = fsc_edges[
        fsc_distance_summary["fsc_bin"].to_numpy(dtype=int) + 1
    ]

    fig, ax = plt.subplots(figsize=(10, 5.5))
    cmap = plt.get_cmap("viridis")
    norm = plt.Normalize(0, 100)
    distances = distance_summary["distance_m"].to_numpy()

    for fsc_bin, group in fsc_distance_summary.groupby("fsc_bin", sort=True):
        lo, hi = fsc_edges[int(fsc_bin):int(fsc_bin) + 2]
        colour = cmap(norm((lo + hi) / 2))
        curve = group.set_index("distance_bin").reindex(distance_summary.index)
        contributing = outer_cells.loc[outer_cells["fsc_bin"] == fsc_bin]
        n_scenes = contributing["scene"].nunique()
        n_cameras = contributing["camera"].nunique()
        label = f"{lo:g} < FSC ≤ {hi:g}% ({n_scenes} scenes, {n_cameras} sites)"
        ax.plot(
            distances, curve["mean_difference_pp"], "o-",
            color=colour, ms=4, lw=2, label=label,
        )
        if SHOW_FSC_SPREAD:
            ax.fill_between(
                distances, curve["lower_pp"].to_numpy(),
                curve["upper_pp"].to_numpy(), color=colour, alpha=0.12,
            )

    ax.plot(
        distances, distance_summary["mean_difference_pp"],
        "--", color="black", lw=2, label="All FSC (pooled mean)",
    )
    ax.axhline(0, color="grey", lw=0.8)
    ax.set(
        title="Distance relationship by observed FSC — all sites pooled",
        xlabel="Distance from camera (m)",
        ylabel="Cell FSC − central FSC (percentage points)",
    )
    ax.grid(alpha=0.2)
    ax.legend(title="Observed FSC groups", fontsize=8)
    if SHOW_FSC_SPREAD:
        ax.set_title(
            "Distance relationship by observed FSC — all sites pooled\n"
            f"Shading: {PERCENTILES[0]}–{PERCENTILES[1]}th percentiles of cell differences"
        )
    fig.tight_layout()
    plt.show()
```

```python
from pathlib import Path
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import rasterio
from rasterio.warp import reproject, Resampling
from pyproj import Transformer

# Settings
LANDSAT_FOLDER = Path("../scratch/optical/landsat_oursites_L1_GEE/TOA/")
ELEVATION_FOLDER = Path("../scratch/optical/landsat_masking/")
WC_FOLDER = Path("../scratch/optical/WC/")

CELL_SIZE_M = 200
DISTANCE_BIN_M = 200
MAX_DISTANCE_M = None
MAX_ELEVATION_DIFFERENCE_M = 80
MIN_VALID_FRACTION = 0.20
PERCENTILES = (16, 84)

# Defaults to the image centre. Optional overrides: camera: (longitude, latitude).
CAMERA_LONLAT = {}

CAMERAS = [
    "bartlett", "delta_junction", "glees", "hyytiala", "kenttarova",
    "lacclair", "marcell", "old_jack_pine", "oregon", "queens",
    "sodankyla", "tammela", "torgnon", "underc", "underhill",
    "varrio", "willowcreek", "wslcreek",
]
WC_NAME_MAP = {
    "sodankyla": "sodankyla_full",
    "marcell": "marcell_MN",
    "oregon": "oregon_yp",
}

assert CELL_SIZE_M > 0 and DISTANCE_BIN_M > 0
assert 0 <= MIN_VALID_FRACTION <= 1

landsat_files = sorted(LANDSAT_FOLDER.glob("*.tif"))
if not landsat_files:
    raise FileNotFoundError(f"No Landsat files found in {LANDSAT_FOLDER}")

elevation_files = list(ELEVATION_FOLDER.iterdir())
cell_results = []

for filepath in landsat_files:
    # Observed FSC is encoded in the filename, as in your original script.
    match = re.search(rf"({'|'.join(CAMERAS)})(\d+)_", filepath.name)
    if match is None:
        continue

    camera = match.group(1)
    observed_fsc = int(match.group(2))

    if not 0 < observed_fsc < 100:
        continue

    elevation_matches = [
        p for p in elevation_files
        if camera in p.name
        and "elevation" in p.name.lower()
        and p.suffix.lower() == ".tif"
    ]
    if len(elevation_matches) != 1:
        raise ValueError(
            f"{camera}: expected one elevation file, "
            f"found {len(elevation_matches)}."
        )

    wc_path = WC_FOLDER / f"{WC_NAME_MAP.get(camera, camera)}.tif"
    print(f"Processing {filepath.name}")

    with rasterio.open(filepath) as src:
        if (
            src.crs is None
            or not src.crs.is_projected
            or not np.isclose(src.crs.linear_units_factor[1], 1)
        ):
            raise ValueError(f"{filepath.name}: CRS must use projected metres.")

        transform = src.transform
        bounds = src.bounds
        height, width = src.height, src.width

        if (
            transform.b != 0 or transform.d != 0
            or transform.a <= 0 or transform.e >= 0
        ):
            raise ValueError(f"{filepath.name}: expected a north-up raster.")

        if CELL_SIZE_M < max(transform.a, -transform.e):
            raise ValueError("CELL_SIZE_M must be at least the pixel size.")

        if camera in CAMERA_LONLAT:
            cx, cy = Transformer.from_crs(
                "EPSG:4326", src.crs, always_xy=True
            ).transform(*CAMERA_LONLAT[camera])
        else:
            cx = (bounds.left + bounds.right) / 2
            cy = (bounds.bottom + bounds.top) / 2

        row, col = src.index(cx, cy)
        if not (0 <= row < height and 0 <= col < width):
            raise ValueError(f"{camera}: camera is outside {filepath.name}.")

        # Green, NIR and SWIR1 bands; reflectance is already scaled.
        if "Landsat7" in filepath.name:
            bands = [2, 4, 5]
        elif "Landsat8" in filepath.name:
            bands = [3, 5, 6]
        else:
            raise ValueError(f"Unknown sensor: {filepath.name}")

        green, nir, swir = (
            src.read(bands, masked=True)
            .astype("float32")
            .filled(np.nan)
        )

        # Reproject forest and elevation layers onto the Landsat grid.
        masks = {}
        for name, path, method in [
            ("forest", wc_path, Resampling.nearest),
            ("elevation", elevation_matches[0], Resampling.bilinear),
        ]:
            aligned = np.full((height, width), np.nan, dtype="float32")

            with rasterio.open(path) as mask_src:
                source = (
                    mask_src.read(1, masked=True)
                    .astype("float32")
                    .filled(np.nan)
                )
                reproject(
                    source=source,
                    destination=aligned,
                    src_transform=mask_src.transform,
                    src_crs=mask_src.crs,
                    src_nodata=np.nan,
                    dst_transform=transform,
                    dst_crs=src.crs,
                    dst_nodata=np.nan,
                    resampling=method,
                )

            masks[name] = aligned
            del source

    centre_elevation = masks["elevation"][row, col]
    if not np.isfinite(centre_elevation):
        print("  Skipping: missing camera elevation.")
        continue

    # Your original Dozier classification.
    denominator = green + swir
    ndsi = np.divide(
        green - swir,
        denominator,
        out=np.full_like(green, np.nan),
        where=denominator != 0,
    )
    snow = (
        ((ndsi > 0.1) & (ndsi < 0.4))
        | ((ndsi >= 0.4) & (nir > 0.11))
    )

    valid = (
        np.isfinite(ndsi)
        & np.isfinite(nir)
        & (masks["forest"] == 10)
        & (
            np.abs(masks["elevation"] - centre_elevation)
            <= MAX_ELEVATION_DIFFERENCE_M
        )
    )

    # Assign pixels to non-overlapping cells, one centred on the camera.
    x = transform.c + (np.arange(width) + 0.5) * transform.a
    y = transform.f + (np.arange(height) + 0.5) * transform.e

    gx = np.floor((x - cx) / CELL_SIZE_M + 0.5).astype(int)
    gy = np.floor((y - cy) / CELL_SIZE_M + 0.5).astype(int)

    nx = gx.max() - gx.min() + 1
    ny = gy.max() - gy.min() + 1
    cell_id = (
        (gy[:, None] - gy.min()) * nx
        + (gx[None, :] - gx.min())
    )

    total_count = np.bincount(cell_id.ravel(), minlength=nx * ny)
    valid_count = np.bincount(cell_id[valid], minlength=nx * ny)
    snow_count = np.bincount(cell_id[valid & snow], minlength=nx * ny)

    grid_x, grid_y = np.meshgrid(
        np.arange(gx.min(), gx.max() + 1),
        np.arange(gy.min(), gy.max() + 1),
    )

    cells = pd.DataFrame({
        "grid_x": grid_x.ravel(),
        "grid_y": grid_y.ravel(),
        "n_valid_pixels": valid_count,
        "valid_fraction": valid_count / np.maximum(total_count, 1),
        "fsc_pct": np.divide(
            100.0 * snow_count,
            valid_count,
            out=np.full(nx * ny, np.nan),
            where=valid_count > 0,
        ),
    })

    cells["x"] = cx + cells["grid_x"] * CELL_SIZE_M
    cells["y"] = cy + cells["grid_y"] * CELL_SIZE_M
    cells["distance_m"] = CELL_SIZE_M * np.hypot(
        cells["grid_x"], cells["grid_y"]
    )

    # Retain complete cells with enough valid forest pixels.
    half = CELL_SIZE_M / 2
    keep = (
        (cells["x"] - half >= bounds.left)
        & (cells["x"] + half <= bounds.right)
        & (cells["y"] - half >= bounds.bottom)
        & (cells["y"] + half <= bounds.top)
        & (cells["valid_fraction"] >= MIN_VALID_FRACTION)
        & cells["fsc_pct"].notna()
    )
    if MAX_DISTANCE_M is not None:
        keep &= cells["distance_m"] <= MAX_DISTANCE_M

    cells = cells.loc[keep].copy()

    # Release full-scene arrays before processing the next scene.
    del green, nir, swir, denominator, ndsi, snow, valid
    del masks, aligned, cell_id

    centre = cells.loc[
        (cells["grid_x"] == 0) & (cells["grid_y"] == 0), "fsc_pct"
    ]
    if centre.empty:
        print("  Skipping: no usable central FSC cell.")
        continue

    cells["centre_fsc_pct"] = centre.iloc[0]
    cells["difference_pp"] = cells["fsc_pct"] - centre.iloc[0]
    cells["observed_fsc_pct"] = observed_fsc
    cells["camera"] = camera
    cells["scene"] = filepath.name
    cell_results.append(cells)

if not cell_results:
    raise ValueError("No partial-snow scenes with usable central FSC cells.")

cells_df = pd.concat(cell_results, ignore_index=True)
del cell_results

# Plot 1: one point per scene, coloured by location.
scenes_df = cells_df[
    ["scene", "camera", "observed_fsc_pct", "centre_fsc_pct"]
].drop_duplicates("scene").sort_values(["camera", "scene"]).reset_index(drop=True)
scenes_df["scene_id"] = [f"S{i + 1:03d}" for i in range(len(scenes_df))]
scene_ids = scenes_df.set_index("scene")["scene_id"].to_dict()
print("\nScene key:")
print(scenes_df[["scene_id", "camera", "observed_fsc_pct", "scene"]].to_string(index=False))

fig, ax = plt.subplots(figsize=(6, 6))
locations = sorted(scenes_df["camera"].unique())
colours = plt.get_cmap("tab20", len(locations))

for i, location in enumerate(locations):
    subset = scenes_df.loc[scenes_df["camera"] == location]
    ax.scatter(
        subset["observed_fsc_pct"],
        subset["centre_fsc_pct"],
        color=colours(i),
        label=location,
        s=40,
        alpha=0.8,
        edgecolors="none",
    )

for record in scenes_df.itertuples():
    ax.annotate(
        record.scene_id, (record.observed_fsc_pct, record.centre_fsc_pct),
        xytext=(4, 4), textcoords="offset points", fontsize=7,
    )

ax.plot([0, 100], [0, 100], "--", color="grey", lw=1)
ax.set(
    xlabel="Observed FSC (%)",
    ylabel="Landsat FSC in camera-centred cell (%)",
    xlim=(-2, 102),
    ylim=(-2, 102),
)
ax.set_aspect("equal")
ax.grid(alpha=0.2)
ax.legend(
    title="Location", bbox_to_anchor=(1.02, 1), loc="upper left"
)
fig.tight_layout()
plt.show()

# Plot 2: pool differences across all scenes, giving each cell equal weight.
# Exclude central cells, whose differences are zero by definition.
outer_cells = cells_df.loc[cells_df["distance_m"] > 0].copy()
distance_summary = pd.DataFrame()
scene_distance_summary = pd.DataFrame()

if outer_cells.empty:
    print("No usable cells outside the central cells.")
else:
    outer_cells["distance_bin"] = (
        outer_cells["distance_m"] // DISTANCE_BIN_M
    ).astype(int)

    distance_summary = (
        outer_cells.groupby("distance_bin")
        .agg(
            mean_difference_pp=("difference_pp", "mean"),
            lower_pp=(
                "difference_pp",
                lambda v: np.percentile(v, PERCENTILES[0]),
            ),
            upper_pp=(
                "difference_pp",
                lambda v: np.percentile(v, PERCENTILES[1]),
            ),
            n_cells=("difference_pp", "size"),
            n_scenes=("scene", "nunique"),
        )
    )

    # Preserve empty bins as gaps in the plot.
    distance_summary = distance_summary.reindex(
        range(int(outer_cells["distance_bin"].max()) + 1)
    )
    distance_summary["distance_m"] = (
        distance_summary.index + 0.5
    ) * DISTANCE_BIN_M

    # Each scene contributes one mean per distance bin. The pooled curve
    # above still gives each cell equal weight, as in the original script.
    scene_distance_summary = (
        outer_cells.groupby(["camera", "scene", "distance_bin"])
        .agg(
            mean_difference_pp=("difference_pp", "mean"),
            n_cells=("difference_pp", "size"),
        )
        .reset_index()
    )
    scene_distance_summary["scene_id"] = (
        scene_distance_summary["scene"].map(scene_ids)
    )
    scene_distance_summary["distance_m"] = (
        scene_distance_summary["distance_bin"] + 0.5
    ) * DISTANCE_BIN_M
    total_cells = scene_distance_summary.groupby("distance_bin")["n_cells"].transform("sum")
    scene_distance_summary["weight_pct"] = (
        100 * scene_distance_summary["n_cells"] / total_cells
    )
    scene_distance_summary["contribution_pp"] = (
        scene_distance_summary["mean_difference_pp"]
        * scene_distance_summary["weight_pct"] / 100
    )
    # Within each bin, contributions sum to the original pooled mean.
    observed_by_scene = scenes_df.set_index("scene")["observed_fsc_pct"].to_dict()
    fsc_cmap = plt.get_cmap("viridis")
    fsc_norm = plt.Normalize(0, 100)
    camera_markers = dict(zip(
        locations,
        ["o", "s", "^", "v", "<", ">", "D", "d", "p", "h", "H",
         "*", "P", "X", "+", "x", "1", "2"],
    ))
    bins = distance_summary.index
    distances = distance_summary["distance_m"].to_numpy()

    fig, ax = plt.subplots(figsize=(12, 6), layout="constrained")
    ax.fill_between(
        distances, distance_summary["lower_pp"].to_numpy(),
        distance_summary["upper_pp"].to_numpy(),
        color="grey", alpha=0.15,
    )
    for (camera, scene), group in scene_distance_summary.groupby(["camera", "scene"]):
        curve = group.set_index("distance_bin").reindex(bins)
        ax.plot(
            distances, curve["mean_difference_pp"],
            color=fsc_cmap(fsc_norm(observed_by_scene[scene])),
            lw=1.2, marker=camera_markers[camera], ms=5,
            markeredgewidth=0.8,
            label=f"{scene_ids[scene]}: {camera}, {observed_by_scene[scene]:g}%",
        )
    ax.plot(
        distances, distance_summary["mean_difference_pp"],
        color="black", lw=2.5, label="Pooled mean (equal cell weights)",
    )
    handles = [
        Line2D(
            [0], [0], color="0.35", marker=camera_markers[camera],
            linestyle="none", markersize=6, label=camera,
        )
        for camera in locations
    ]
    handles += [
        Line2D([0], [0], color="black", lw=2.5, label="Pooled mean"),
        Line2D([0], [0], color="grey", lw=6, alpha=0.3,
               label=f"Pooled {PERCENTILES[0]}–{PERCENTILES[1]}th percentiles"),
    ]
    ax.axhline(0, color="black", lw=0.8)
    ax.set(
        title="Individual scenes: colour = observed FSC; marker = camera",
        xlabel="Distance from camera (m)",
        ylabel="Cell FSC − central FSC (percentage points)",
    )
    ax.grid(alpha=0.2)
    ax.legend(
        handles=handles, title="Camera / pooled statistics",
        bbox_to_anchor=(1.02, 1), loc="upper left", fontsize=8,
    )
    fig.colorbar(
        plt.cm.ScalarMappable(norm=fsc_norm, cmap=fsc_cmap),
        ax=ax, orientation="horizontal", label="Observed FSC (%)",
        ticks=np.arange(0, 101, 20), fraction=0.07, pad=0.08,
    )
    plt.show()

    # Camera panels identify individual scenes using the same IDs as Plot 1.
    # Paginate to keep legends readable when there are many scenes.
    scene_distance_summary["scene_id"] = scene_distance_summary["scene"].map(scene_ids)
    panels = []
    for camera, group in scene_distance_summary.groupby("camera", sort=True):
        scene_names = sorted(group["scene"].unique())
        for offset in range(0, len(scene_names), 8):
            panels.append((camera, scene_names[offset:offset + 8]))


    for page_start in range(0, len(panels), 6):
        page = panels[page_start:page_start + 6]
        ncols = min(2, len(page))
        nrows = int(np.ceil(len(page) / ncols))
        fig, axes = plt.subplots(
            nrows, ncols, figsize=(7 * ncols, 4 * nrows),
            sharex=True, sharey=True, squeeze=False, layout="constrained",
        )
        for ax, (camera, scene_names) in zip(axes.flat, page):
            for scene in scene_names:
                group = scene_distance_summary.loc[scene_distance_summary["scene"] == scene]
                curve = group.set_index("distance_bin").reindex(bins)
                ax.plot(
                    distances, curve["mean_difference_pp"],
                    color=fsc_cmap(fsc_norm(observed_by_scene[scene])),
                    marker=".", ms=4, lw=1.4,
                    label=f"{scene_ids[scene]} ({observed_by_scene[scene]:g}%)",
                )
            ax.plot(
                distances, distance_summary["mean_difference_pp"],
                color="black", ls="--", lw=2, label="Pooled mean",
            )
            ax.axhline(0, color="grey", lw=0.8)
            ax.set_title(camera)
            ax.set_xlabel("Distance from camera (m)")
            ax.set_ylabel("Cell FSC − central FSC (pp)")
            ax.grid(alpha=0.2)
            ax.legend(fontsize=8, ncol=3)
        for ax in list(axes.flat)[len(page):]:
            ax.set_visible(False)
        fig.colorbar(
            plt.cm.ScalarMappable(norm=fsc_norm, cmap=fsc_cmap),
            ax=list(axes.flat)[:len(page)],
            label="Observed FSC (%)", shrink=0.85, pad=0.03,
        )
        plt.show()

    print("\nPer-scene contributions are available in scene_distance_summary:")
    print("weight_pct = percentage of cells in that distance bin supplied by the scene.")
    print("contribution_pp = scene mean × weight; sums to the pooled mean in each bin.")
```
